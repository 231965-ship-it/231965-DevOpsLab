```python
import pandas as pd

# Added on_bad_lines='skip' to bypass malformed rows causing the ParserError
df = pd.read_csv("/college_major_career_outcomes_2026.csv", on_bad_lines='skip')

print(df.head())
print(df.shape)
print(df.info())
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values[missing_values > 0])
```

```python
# Drop rows where 'linked_occupations_count' is missing
df.dropna(subset=['linked_occupations_count'], inplace=True)

print(f"DataFrame shape after dropping missing values in 'linked_occupations_count': {df.shape}")
```

```python
# Display summary statistics for relevant earnings columns
print(df[['median_earnings_4yr_usd', 'median_earnings_1yr_usd', 'median_earnings_5yr_usd']].describe())
```

```python
# Fill missing values in object columns with 'unknown'
for column in df.columns:
    if df[column].dtype == 'object':
        if df[column].isnull().any():
            df[column].fillna('unknown', inplace=True)

print("Missing values after filling 'object' columns with 'unknown':")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
for column in df.select_dtypes(include='object').columns:
    df[column].fillna(df[column].mode()[0], inplace=True)
```

```python
print("Duplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)
```

```python
print(df.dtypes)
```

```python
from sklearn.preprocessing import StandardScaler

numerical_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns

scaler = StandardScaler()

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)
```

```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df = df[
        (df[column] >= lower_limit) &
        (df[column] <= upper_limit)
    ]
```

```python
df.to_csv("preprocessed_dataset.csv", index=False)

print("Dataset saved successfully!")
print("Final shape:", df.shape)
```

```python
# Re-load the original dataset to reset the DataFrame with error handling for malformed rows
df = pd.read_csv("/college_major_career_outcomes_2026.csv", on_bad_lines='skip', low_memory=False)

print(f"DataFrame shape after re-loading: {df.shape}")
```

```python
# Drop rows where 'linked_occupations_count' is missing (as previously done)
df.dropna(subset=['linked_occupations_count'], inplace=True)

print(f"DataFrame shape after dropping missing values in 'linked_occupations_count': {df.shape}")
```

```python
# Fill missing values in object columns with their mode (as previously done)
for column in df.select_dtypes(include='object').columns:
    df[column].fillna(df[column].mode()[0], inplace=True)

print("Missing values in object columns filled with mode.")
```

```python
# Drop duplicate rows (as previously done)
initial_rows = df.shape[0]
df.drop_duplicates(inplace=True)
final_rows = df.shape[0]

print(f"Duplicates found and removed: {initial_rows - final_rows}")
print(f"DataFrame shape after dropping duplicates: {df.shape}")
```

```python
# Display remaining missing values
missing_values_after_cleaning = df.isnull().sum()
print("Missing values per column after initial cleaning (excluding outlier removal):")
print(missing_values_after_cleaning[missing_values_after_cleaning > 0])
```

### Imputing numerical missing values with the median

```python
# Identify numerical columns with missing values
numerical_cols_with_missing = df.select_dtypes(include=['int64', 'float64']).columns

# Impute missing values with the median for each numerical column
for col in numerical_cols_with_missing:
    if df[col].isnull().any():
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)

print("Numerical missing values imputed with the median.")
```

```python
# Verify no more missing values exist
missing_values_final = df.isnull().sum()
print("Remaining missing values after all imputation:")
print(missing_values_final[missing_values_final > 0])

print(f"Final DataFrame shape: {df.shape}")
```

### Identifying Target Variable and Features for Linear Regression

```python
# Define the target variable
target_variable = 'median_earnings_4yr_usd'

# Identify potential features (all numerical columns excluding the target and identifiers)
# Exclude 'program_id', 'unitid', 'opeid6' as they are identifiers
# Exclude other earnings-related columns if they are highly correlated or also targets

numerical_features = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Remove target variable and other non-feature columns from the list
exclude_columns = [
    target_variable,
    'unitid',
    'opeid6',
    'median_earnings_1yr_usd',
    'median_earnings_5yr_usd',
    'earnings_cohort_size_1yr',
    'earnings_cohort_size_4yr',
    'earnings_cohort_size_5yr'
]

features = [col for col in numerical_features if col not in exclude_columns]

X = df[features] # Features DataFrame
y = df[target_variable] # Target Series

print(f"Target Variable: {target_variable}")
print(f"Number of Features: {len(features)}")
print("First 5 feature names:")
print(features[:5])

print("\nShape of X (features):", X.shape)
print("Shape of y (target):", y.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Initialize the Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Display the model's intercept and coefficients
print(f"Model Intercept: {model.intercept_}")
print("\nModel Coefficients:")
coefficients_df = pd.DataFrame({'Feature': X.columns, 'Coefficient': model.coef_})
print(coefficients_df.sort_values(by='Coefficient', ascending=False).head())
print(coefficients_df.sort_values(by='Coefficient', ascending=True).head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Make predictions using the trained model
y_pred = model.predict(X)

# Create a scatter plot of actual vs. predicted values
plt.figure(figsize=(10, 8))
sns.scatterplot(x=y, y=y_pred, alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2) # Diagonal line for perfect prediction
plt.xlabel('Actual Median Earnings (4yr USD)')
plt.ylabel('Predicted Median Earnings (4yr USD)')
plt.title('Actual vs. Predicted Median Earnings (4yr USD)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert y to numeric values (floats) so we can subtract y_pred
y_numeric = pd.to_numeric(y, errors='coerce')

# Calculate residuals
residuals = y_numeric - y_pred

# Plot the distribution of residuals
plt.figure(figsize=(10, 6))
sns.histplot(residuals.dropna(), kde=True)
plt.title('Distribution of Residuals')
plt.xlabel('Residuals (Actual - Predicted Earnings)')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
